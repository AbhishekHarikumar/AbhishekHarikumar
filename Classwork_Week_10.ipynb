```python
# Import the necessary libraries

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Q1. Line Plot and Markers
# Create a line plot using the following data:
# x = [1, 2, 3, 4, 5, 6]
# y = [4, 9, 7, 12, 10, 16]
# Use a blue dashed line, circular markers of size 8, a title, axis labels, and gridlines. Explain whether the values increase consistently.

x = [1, 2, 3, 4, 5, 6]
y = [4, 9, 7, 12, 10, 16]

plt.plot(
    x,
    y,
    color = 'blue',
    marker = 'o',
    markersize = 8,
    linestyle = '--'
  )
plt.title("Line Graph")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Q2. Multiple Lines in One Graph
# Given:
# days = [1, 2, 3, 4, 5, 6]
# product_A = [12, 15, 18, 17, 22, 25]
# product_B = [10, 14, 13, 19, 20, 23]
# Plot both products in the same figure using different colours, markers, and line styles. Include a legend. Identify the days when Product B exceeds Product A.

days = [1, 2, 3, 4, 5, 6]
product_A = [12, 15, 18, 17, 22, 25]
product_B = [10, 14, 13, 19, 20, 23]

plt.plot(days, product_A, marker = 'o', markersize = 8, label = 'Product A', linestyle = '--', color = 'green')
plt.plot(days, product_B, marker = 'o', markersize = 6, label = 'Product B', linestyle = '--', color = 'grey')
plt.grid(True)
plt.title("Multi Line Graph")
plt.xlabel("Days")
plt.ylabel("Product values")
plt.legend()
plt.show()

# On Day 4, Product B exceeds Product A
```

```python
# Q3. Understanding Subplots
# Generate the following values:
# x = np.arange(1, 11)
# y1 = x**2
# y2 = x**3
# Create two subplots placed side by side. Display a line plot of y1 in the first subplot and a scatterplot of y2 in the second. Add separate titles and axis labels.

x = np.arange(1, 11)
y1 = x**2
y2 = x**3

figure, (ax, bx) = plt.subplots(1,2,figsize = (10,5))
ax.plot(x, y1, color = 'pink')
ax.grid()
ax.set_xlabel("x")
ax.set_ylabel("y1")
ax.set_title("Line plot")

bx.scatter(x, y2, color = 'green')
bx.grid()
bx.set_xlabel('x')
bx.set_ylabel('y2')
bx.set_title("Scatter plot")
plt.show()
```

```python
# Q4. Changing Marker Sizes and Transparency
# Given:
# x = [1, 2, 3, 4, 5]
# y = [10, 15, 12, 20, 18]
# sizes = [50, 100, 150, 200, 250]
# Create a scatterplot where each point has a different size using sizes. Set transparency to 0.5. Explain how marker size and transparency affect the visualization.

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]
sizes = [50, 100, 150, 200, 250]

plt.scatter(x, sizes, color = 'black', alpha = 0.5, s = 50)
plt.scatter(y, sizes, color = 'Green', alpha = 0.4, s = 150)
plt.grid(True)
plt.title("Scatter Plot")
plt.xlabel("x, y")
plt.ylabel("Sizes")
plt.show()
```

```python
# Q5. Assigning Colours Using a Colormap
# Create a scatterplot using:
# x = [2, 4, 6, 8, 10]
# y = [5, 9, 7, 14, 18]
# values = [10, 30, 50, 70, 90]
# Assign colors using the viridis colormap based on values. Add a colorbar. Repeat the graph using viridis_r. Explain the difference.

x = [2, 4, 6, 8, 10]
y = [5, 9, 7, 14, 18]
values = [10, 30, 50, 70, 90]

figure, (cx, dx) = plt.subplots(1,2,figsize = (10,5))
# With Virdis
cx.scatter(x, values, c = range(len(x)),  cmap = 'viridis', s = 100, alpha = 0.6, label = 'X')
cx.scatter(y, values, c = range(len(y)), cmap = 'viridis', s = 50, alpha = 0.4, label = 'Y')
cx.scatter(y, values)
cx.scatter(y, values)
cx.grid()
cx.set_xlabel("X, Y")
cx.set_ylabel("Values")
# cx.colorbar(sm, ax = cx)
cx.legend()
cx.set_title("Scatter Plot with Virdis")

# With Virdis_r
dx.scatter(x, values, c = range(len(x)),  cmap = 'viridis', s = 100, alpha = 0.6, label = 'X')
dx.scatter(y, values, c = range(len(y)), cmap = 'viridis', s = 50, alpha = 0.4, label = 'Y')
dx.scatter(y, values)
dx.scatter(y, values)
dx.grid()
dx.set_xlabel("X, Y")
dx.set_ylabel("Values")
dx.legend()
dx.set_title("Scatter Plot with Virdis_r")
plt.show()
```
